# Skills-gap extension
Skills gap, as measurable from the LFS: the difference in education and vocational-training attainment between unemployed and employed youth, the education mismatch of employed youth relative to the occupations they hold, and what happens to youth at each education level in the labour market. Education and training are proxies for skills. Vacancies and employer demand are not measured in the LFS; employer surveys (NAVTTC/NSIS) are cited as context only.

## Survey data and approved definitions
See the audit and methods for routing, weights, exclusions and limitations.

```python
import pandas as pd
from pathlib import Path
tables = Path('../reports/tables')
pd.read_csv(tables / 'skills_reproduction.csv')[['indicator', 'difference_pp', 'status']]
```

## Weighted survey findings
- The training share among unemployed youth was -8.9 percentage points compared with employed youth.
- Under the all-employed norm, 46.0% of employed youth were above their occupation group's modal education band.
- Using the youth-only norm, the corresponding share was 41.0%.
- The approximate not-employed-and-not-enrolled share was 29.3% of all youth.
- Among non-enrolled youth in the labour force reporting another two-year degree, 25.1% were unemployed.
The model-estimated education-profile difference averaged +3.6 percentage points among covered profiles; weighted coverage was 82.1%. This is a model-estimated difference between profiles, not a causal effect.
